#### 12.beverages ürün kategorisindeki ürünlerin fiyat ortalaması nedir ?

```python
%%sql
SELECT CA.CategoryName AS Kategori, PR.ProductName AS Urun_adi, AVG(PR.Price) AS Ortalama_fiyat
FROM Products PR
INNER JOIN CATEGORIES CA
      ON PR.CategoryID = CA.CategoryID
WHERE CategoryName = 'Beverages'
GROUP BY(ProductName);
```

#### 13.USA'da yaşayan müşterilerin kazandırdığı toplam kazancı bulunuz ? 

```python
#13.Cevap(Orders, OrderDetails, Customers, Products)
```

```python
%%sql
SELECT Customers.CustomerName AS Musteri_Adi, Customers.Country AS Ulke, SUM(Products.Price * OrderDetails.Quantity) AS Toplam_Kazanc
FROM Customers
JOIN Orders
    ON Customers.CustomerID = Orders.CustomerID
JOIN OrderDetails
    ON Orders.OrderID = OrderDetails.OrderID
JOIN Products
    ON OrderDetails.ProductID = Products.ProductID
WHERE Customers.Country = 'USA'
GROUP BY(CustomerName);
```

#### 14.Product Tablosundaki CategoryID'lerin isimlerini bulunuz ve sonrasında kategori başına ortalama ürün fiyatını gösteriniz ?

#### 15.Çalışanları isimleriyle birlikte yaptıkları toplam satışlara göre sıralayınız ? 

```python
%%sql
SELECT Em.EmployeeID AS Calisan_ID, Em.FirstName AS Isim, Em.LastName AS Soyisim, SUM(Pr.Price * OrDt.Quantity) AS Toplam_Satis
FROM Employees Em
JOIN Orders Ord 
     ON Em.EmployeeID = Ord.EmployeeID
JOIN OrderDetails OrDt 
     ON Ord.OrderID = OrDt.OrderID
JOIN Products Pr 
     ON OrDt.ProductID = Pr.ProductID
GROUP BY(Ord.EmployeeID)
ORDER BY(Toplam_Satis) DESC;
```

#### 16.Germany'den verilen siparişlerin kategorilerine göre ortalama fiyatları nelerdir ? 

```python
#16.Cevap
```

```python
%%sql
SELECT Categories.CategoryName AS Kategori_Adi, AVG(Products.Price) AS Ortalama_Fiyat
FROM Categories
JOIN Products 
     ON Categories.CategoryID = Products.CategoryID
JOIN OrderDetails 
     ON Products.ProductID = OrderDetails.ProductID
JOIN Orders 
     ON OrderDetails.OrderID = Orders.OrderID
JOIN Customers 
     ON Orders.CustomerID = Customers.CustomerID
WHERE (Customers.Country = 'Germany')
GROUP BY(Categories.CategoryName);
```

#### 17.Almanya'dan veya USA'dan verilen siparişlerin kategorilerine göre ortalama fiyatları nelerdir?

```python
%%sql
SELECT Categories.CategoryName AS Kategori_Adi, AVG(Products.Price) AS Ortalama_Fiyat
FROM Categories
JOIN Products 
     ON Categories.CategoryID = Products.CategoryID
JOIN OrderDetails 
     ON Products.ProductID = OrderDetails.ProductID
JOIN Orders 
     ON OrderDetails.OrderID = Orders.OrderID
JOIN Customers 
     ON Orders.CustomerID = Customers.CustomerID
WHERE (Customers.Country = 'Germany') OR (Customers.Country = 'USA')
GROUP BY(Categories.CategoryName);
```

#### 18.Haziran-Temmuz-Ağustos aylarında verilen siparişlerin ortalama fiyatı nedir ? 

```python
#18.Cevap
```

```python
%%sql
SELECT AVG(Products.Price * OrderDetails.Quantity) AS Ortalama_fiyat
FROM Products
JOIN OrderDetails 
    ON Products.ProductID = OrderDetails.ProductID
JOIN Orders 
    ON OrderDetails.OrderID = Orders.OrderID
WHERE Orders.OrderDate LIKE '%.06.%'
   OR Orders.OrderDate LIKE '%.07.%'
   OR Orders.OrderDate LIKE '%.08.%';
```

#### 19.Müşterilerin 1997 yılına ait siparişlerinin maksimum miktarlarını bulup müşterilerin isimleri ile birlikte büyükten küçüğe doğru sıralayınız ? 

```python
#19.Cevap
```

```python
%%sql
SELECT Customers.CustomerName AS Musteri_adi, MAX(Products.Price * OrderDetails.Quantity) AS Verilen_en_pahali_siparis_tutari
FROM Customers
JOIN Orders 
       ON Customers.CustomerID = Orders.CustomerID
JOIN OrderDetails 
       ON Orders.OrderID = OrderDetails.OrderID
JOIN Products 
       ON OrderDetails.ProductID = Products.ProductID
WHERE Orders.OrderDate LIKE '%.1997'
GROUP BY(Customers.CustomerName)
ORDER BY(Verilen_en_pahali_siparis_tutari) DESC;
```

#### 20.Sipariş yılı 1997 olan siparişleri alan çalışanları sipariş alma sayılarına göre sıralayınız ? 

```python
#20.Cevap
```

```python
%%sql
SELECT Employees.FirstName AS ad, Employees.LastName AS Soyad, SUM(OrderDetails.Quantity) AS Alınan_siparis_sayisi
FROM Employees
JOIN Orders 
      ON Employees.EmployeeID = Orders.EmployeeID
JOIN OrderDetails 
      ON Orders.OrderID = OrderDetails.OrderID
WHERE OrderDate LIKE '%.1997%'
GROUP BY(Employees.EmployeeID)
ORDER BY(Alınan_siparis_sayisi) DESC;
```
