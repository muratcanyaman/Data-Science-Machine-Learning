#### LIKE ile SEÇİM İŞLEMLERİ

- *'a%': "a" ile başlayan değerler*
- *'%a': "a" ile biten ifadeler*
- *'%or%': Herhangi bir yerinde "or" olan ifadeler*
- *'_r%': İkinci sırasında "r" olan değerler*
- *'a%%': "a" ile başlayıp en az 3 karakterden oluşan değerler*
- *'a%0': "a" ile başlayıp "o" ile biten değerler*

```python
#icinde "or" ifadesi gecen musterilerin cekilmesi
```

```python
%%sql
SELECT *
FROM Customers
WHERE CustomerName LIKE '%or%'
LIMIT 10;
```

```python
#ismi "a" harfi ile biten musteriler
```

```python
%%sql
SELECT *
FROM Customers
WHERE CustomerName LIKE '%a'
LIMIT 10;
```

```python
#ismi 'A' harfi ile baslayan, yasadigi sehir adi 'o' ile biten
```

```python
%%sql
SELECT *
FROM Customers
WHERE CustomerName LIKE 'A%'
AND Country LIKE '%o';
```

```python
#left join
```

```python
%%sql
SELECT *
FROM Products
LEFT OUTER JOIN Categories
ON Products.CategoryID = Categories.CategoryID
LIMIT 5;
```

```python
#Sadece hangi urun hangi kategoriye ait sorusuna cevap bulmak istersek:
```

```python
%%sql
SELECT PR.ProductName, CA.CategoryID, CA.CategoryName
FROM Products PR
LEFT OUTER JOIN Categories CA
ON PR.CategoryID = CA.CategoryID
LIMIT 10;
```
