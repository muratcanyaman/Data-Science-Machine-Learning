## Veri Bilimi Proje Döngüsü

<div align="center"></div>

*Bir Veri Bilimi projesine başlamadan önce cevaplanması gereken birtakım sorular vardır:*

- *İlgili birimlerin veya iş yerinin ihtiyacı nedir ?*
- *Bu proje ile hangi problemi çözeceğiz ? (Tersten gidiş)*
- *Bu işin somut çıktısı ve işe katkısı nedir ?*
- *Neyi ölçmek istiyoruz ?*
- *Neyi tahmin etmek istiyoruz ?*
- *Bu modeli veya tahmini kim kullanacak ?*
- *Projenin başarı metriği nedir ?*
- *Bu işin doğrulamasından ve kontolünden kim sorumlu olacak ?*
- *Bu proje ne kadar sürecek?*
- *Hangi departmanlar hangi seviyelerden işin içinde olacak ?*

*Bir Veri Bilimi Projesi'nde her zaman bir çıktı elde ederiz. Bu çıktının ne kadar doğru, başarılı bir çıktı olduğu da başlangıçta iş problemini, karşı tarafın bizden tam olarak ne beklediğini ne kadar doğru bir şekilde anladığımıza bağlıdır. O yüzden bu alanda proje geliştirmek ve Deploy etmekten çok başlangıçta problemi ne kadar iyi analiz etmişiz, ne kadar doğru anlamışız... Bu sorulara yanıt bulmak daha da önemli hale geliyor.*
